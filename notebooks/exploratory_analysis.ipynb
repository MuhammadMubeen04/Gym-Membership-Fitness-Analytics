# Gym Membership & Fitness Analytics – Exploratory Analysis

Quick EDA notebook for members and check-ins data.

## 1. Imports & Load

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

members = pd.read_csv('../data/raw/members.csv', parse_dates=['JoinDate'])
checkins = pd.read_csv('../data/raw/checkins.csv', parse_dates=['CheckInDate'])
print(members.shape, checkins.shape)
members.head()
```

## 2. Members Overview

```python
print(members['Status'].value_counts())
print(members['MembershipPlan'].value_counts())
members.describe()
```

## 3. Status & Plans

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
members['Status'].value_counts().plot.pie(autopct='%1.1f%%', ax=axes[0])
axes[0].set_title('Membership Status')
members['MembershipPlan'].value_counts().plot(kind='bar', ax=axes[1], color='steelblue')
axes[1].set_title('Members by Plan')
plt.tight_layout()
plt.show()
```

## 4. Class Popularity

```python
checkins['ClassAttended'].value_counts().sort_values().plot(kind='barh', color='#4e79a7')
plt.title('Class / Activity Popularity')
plt.xlabel('Visits')
plt.tight_layout()
plt.show()
```

## 5. Peak Hours

```python
checkins.groupby('CheckInHour').size().plot(kind='bar', color='#f28e2b')
plt.title('Check-ins by Hour')
plt.xlabel('Hour')
plt.ylabel('Check-ins')
plt.tight_layout()
plt.show()
```

## 6. Quick Insights

```python
print('Total members:', len(members))
print('Active rate: {:.1f}%'.format((members['Status']=='Active').mean()*100))
print('Total check-ins:', len(checkins))
print('Avg duration (min):', round(checkins['DurationMinutes'].mean(), 1))
print('Top class:', checkins['ClassAttended'].value_counts().idxmax())
```
